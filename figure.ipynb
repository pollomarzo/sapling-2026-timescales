```python
#| label: fig-tau
#| caption: Intrinsic timescale recovered from each regions autocorrelation, against hierarchical level. Points are the across-seed mean; the band is one standard deviation. Hover a point for its values.

import altair as alt
import pandas as pd

df = pd.DataFrame({
    'level': range(1, 9),
    'tau':  [3, 5, 8, 12, 17, 24, 32, 40],
    'lo':   [2.2, 4.0, 6.8, 10.5, 15.0, 21.5, 29.0, 36.5],
    'hi':   [3.8, 6.0, 9.2, 13.5, 19.0, 26.5, 35.0, 43.5],
})

band = alt.Chart(df).mark_area(opacity=0.15, color='#2f6b4f').encode(
    x=alt.X('level:Q', title='Hierarchical level'),
    y=alt.Y('lo:Q', title='Intrinsic timescale (steps)'), y2='hi:Q')
line = alt.Chart(df).mark_line(color='#2f6b4f', strokeWidth=2.5).encode(x='level:Q', y='tau:Q')
points = alt.Chart(df).mark_point(color='#2f6b4f', filled=True, size=70).encode(
    x='level:Q', y='tau:Q', tooltip=['level', 'tau', 'lo', 'hi'])

(band + line + points).properties(width=460, height=280)
```
