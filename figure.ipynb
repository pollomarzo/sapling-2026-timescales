```python
#| label: fig-tau

import altair as alt
rows = [{'level': i+1, 'tau': t, 'lo': l, 'hi': h}
        for i, (t, l, h) in enumerate(zip(
            [3,5,8,12,17,24,32,40], [2.2,4,6.8,10.5,15,21.5,29,36.5], [3.8,6,9.2,13.5,19,26.5,35,43.5]))]
data = alt.Data(values=rows)
band = alt.Chart(data).mark_area(opacity=0.15, color='#2f6b4f').encode(
    x=alt.X('level:Q', title='Hierarchical level'), y=alt.Y('lo:Q', title='Intrinsic timescale (steps)'), y2='hi:Q')
line = alt.Chart(data).mark_line(color='#2f6b4f').encode(x='level:Q', y='tau:Q')
pts = alt.Chart(data).mark_point(color='#2f6b4f', filled=True).encode(x='level:Q', y='tau:Q', tooltip=['level','tau','lo','hi'])
(band + line + pts).properties(width=460, height=280)
```
